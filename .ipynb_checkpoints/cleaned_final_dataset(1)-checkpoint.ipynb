```python
import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer

# 1. Download the VADER lexicon (Run this once)
nltk.download('vader_lexicon')

# 2. Initialize the Analyzer
sia = SentimentIntensityAnalyzer()
```

```python
def robust_sentiment_analysis(text):
    """
    Uses NLTK VADER to calculate sentiment scores.
    Returns: pos, neg, neu, compound
    """
    # 1. Handle Missing Data (NaN)
    if pd.isna(text) or text == "":
        return pd.Series([0.0, 0.0, 0.0, 0.0]) # Return Neutral if no text

    # 2. Ensure string format
    text = str(text)

    # 3. Get Scores
    # VADER returns a dict: {'neg': 0.0, 'neu': 0.0, 'pos': 0.0, 'compound': 0.0}
    scores = sia.polarity_scores(text)

    return pd.Series([
        scores['pos'],      # Positive Component (0 to 1)
        scores['neg'],      # Negative Component (0 to 1)
        scores['neu'],      # Neutral Component (0 to 1)
        scores['compound']  # Normalized Score (-1 to +1) <--- USE THIS FOR ANALYSIS
    ])
```

```python
# Create new columns
print("Calculating sentiment... this may take a moment.")
df[['pos_score', 'neg_score', 'neu_score', 'compound_score']] = df['text_snippet'].apply(robust_sentiment_analysis)

# --- CHECK RESULTS ---
print("Done! Here are the top 5 most positive articles:")
print(df.sort_values(by='compound_score', ascending=False)[['headline', 'compound_score']].head())

print("\nHere are the top 5 most negative articles:")
print(df.sort_values(by='compound_score', ascending=True)[['headline', 'compound_score']].head())
```
